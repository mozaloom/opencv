#Task (3)

Flip the input image horizontally and vertically so it acts like a mirror. You can
try to do this using an OpenCV function or by manually switching each pixel.
Set up your program so the image is flipped depending on a keypress by the user. For example, if the user types 'h' then the image should be shown
horizontally flipped. If the user types 'v', then the image is shown vertically
flipped.


```python
img=cv.imread("images/ROBOT.jpg")
cv2_imshow(img)
```

```python
flag=True
while(flag):
  x=input("Enter 'h' for  horizantal flipping, 'v' for vertical  flipping ")
  if x=='h':
    cv2_imshow(cv.flip(img,1))
    flag=False
  elif x=='v':
    cv2_imshow(cv.flip(img,0))
    flag=False
  else:
    print("wrong index, please try again ")
```

#Task (4)

Pick a filter--such as a Gaussian blur or a Laplacian filter--and apply it to your
input image. Make sure the filter is doing what you expect it to do.

```python
img=cv.imread('images/ROBOT.jpg')
cv2_imshow(img)
KernalSize=[(3,3),(5,5),(17,17)]
for (kx,ky) in KernalSize:
  blurred=cv.GaussianBlur(img,(kx,ky),0)
cv2_imshow(blurred)
```
